```python
# Cody Newkirk
# SDEV220
# This program uses multiprocessing to create three separate student processes, 
# each waiting a random amount of time (0–1 second), printing their actions and the time, and exiting.

import multiprocessing
import time
import random
from datetime import datetime

# Define a student function
def student():
    wait_time = random.random()  # Random wait between 0-1 seconds
    print(f"{multiprocessing.current_process().name} is starting. Will wait for {wait_time:.2f} seconds.")
    time.sleep(wait_time)
    print(f"{multiprocessing.current_process().name} woke up at {datetime.now()}.")
    print(f"{multiprocessing.current_process().name} has finished.\n")

# Create three student processes
if __name__ == '__main__':
    processes = []

    for i in range(3):
        process = multiprocessing.Process(target=student, name=f'Student-{i+1}')
        processes.append(process)
        process.start()

    for process in processes:
        process.join()

    print("All students have completed their tasks!")
```
