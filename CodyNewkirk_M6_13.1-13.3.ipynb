```python
# Cody Newkirk
# SDEV220
# This program writes today's date to a file, reads it back, and parses it into a date object.

from datetime import date, datetime

# Step 13.1: Write the current date as a string to the text file today.txt
today = date.today()
with open("today.txt", "w") as file:
    file.write(today.isoformat())  # Format: YYYY-MM-DD

# Step 13.2: Read the text file today.txt into the string today_string
with open("today.txt", "r") as file:
    today_string = file.read()

# Step 13.3: Parse the date from today_string
parsed_date = datetime.strptime(today_string, "%Y-%m-%d").date()

# Print results
parsed_date
```
